```python
from langchain_core.prompts import ChatPromptTemplate
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
load_dotenv()
```

```python
model = init_chat_model(
     model="ministral-3b-latest",
     model_provider='mistralai'
)
```

# Use of Prompt Template

```python
prompt = ChatPromptTemplate.from_template("""
Extract the following information about the movie from the given text:

- Movie title
- Release year
- Director
- Genre
- Main actors
- Plot summary
- IMDb rating

If any information is missing, return "Not available".
Do not invent information.

Movie information:
{text}
""")

# Example
formatted_prompt = prompt.invoke({
    "text": "Inception is a 2010 sci-fi thriller directed by Christopher Nolan, \
starring Leonardo DiCaprio. It follows a thief who enters people's dreams."
})
```

```python
response = model.invoke(formatted_prompt)
response.content
```

# use of Structured output

```python
from pydantic import BaseModel
```

```python
from typing import List


class Movie(BaseModel):
    title: str 
    release_year : int | None = None
    genre : List[str]
    director : str|None = None
    cast : List[str]
    rating : float|None = None
    Summary : str 
```

```python
structured_model = model.with_structured_output(Movie)
```

```python
prompt = ChatPromptTemplate.from_messages([
    ("system", """
    Extract movie information from the paragraph.
    Use only information supported by the paragraph.
    Do not invent missing details.
    """),
    ("human", "{paragraph}")
])
```

```python
chain = prompt | structured_model
```

```python
user = "Inception is a 2010 sci-fi thriller directed by Christopher Nolan, \
starring Leonardo DiCaprio. It follows a thief who enters people's dreams."
```

```python
result = chain.invoke({
    "paragraph": user
})
```

```python
print(result)
print(result.model_dump())
```
